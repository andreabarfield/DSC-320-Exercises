```python
# Calculating Limits Numerically
# ------------------------------
# a)
# import numpy
import numpy as np

# define the f(x)
def f(x):
    return 2*x**3 - 4*x + 1

# assign numbers close to 3 on either side to "x" variable
# create an array with numpy
x = np.array([2.9, 2.99, 2.999, 3.001, 3.01, 3.1])

# assign function to "y" variable
y = f(x)

# print values of x and y as it approaches the limit
print("x values: ")
print(x)
print("\nf(x) values: ")
print(y)
```

```python
# Calculating Limits Numerically
# ------------------------------
# b)

# define g(x)
def g(x):
    return (np.exp(x) - 1) / x

# assign numbers close to 0 on either side to "x" variable
# create an array with numpy
x = np.array([-0.1, -0.01, -0.001, 0.001, 0.01, 0.1])

# assign function to "y" variable 
y = g(x)

# print values of x and y as it approached the limit
print("x values: ")
print(x)
print("\nf(x) values: ")
print(y)
```

```python
# Average Rate of Change
# ----------------------

# define function that will take in a function 
# and return the average rate of change between a and b.
def average_rate_of_change(f, a , b):
    return (f(b) - f(a)) / (b - a)

# test the function
def f(x):
    return 3 * x**2

# call the average_rate_of_change function using f(x), a = 2, b = 5
average_rate_of_change(f, 2, 5)

    
```

```python
# Average Rate of Change to Instantaneous Rate of Change
# ------------------------------
# a) 

# define the function f(t)
def f(t):
    return 4.9*t**2

# define average_rate_of_change function
def average_rate_of_change(f, a , b):
    return (f(b) - f(a)) / (b - a)

# find average speed at a = 5 and b = 6
speed_5_6 = average_rate_of_change(f, 5, 6)
print(f'a) Average speed from 5-6 seconds: {speed_5_6:.2f} m/s')

# b)
# ------------------------------

# find average speed at a = 5 and b = 5.5
speed_5_55 = average_rate_of_change(f, 5, 5.5)
print(f'b) Average speed from 5-6 seconds: {speed_5_55:.2f} m/s')

# c)
# ------------------------------

# find average speed at a = 5 and b = 5.1
speed_5_51 = average_rate_of_change(f, 5, 5.1)
print(f'c) Average speed from 5-6 seconds: {speed_5_51:.2f} m/s')

# d)
# ------------------------------

# create an array with numpy to check values approaching 5
# assign it to "intervals" variable
intervals = np.array([6, 5.5, 5.1, 5.01, 5.001, 5.0001])

# for loop to plug values into average rate of change formula
for t in intervals:
    speed = average_rate_of_change(f, 5, t)
    print()
    print(f'd) 5 to {t}: {speed:.4f} m/s')

# e) & f)
# ------------------------------

# define derivative of f(t)
def derivative(t):
    return 2 * 4.9 * t 

print()
print(f"f) f'(5) = {derivative(5):.4f} m/s")

# In part d, as b gets closer to 5 the speed gets closer to 49 
# and in part f, the derivative at 5 is 49. Part f supports
# the estimation I found in part d.
```

```python
# Calculating and Interpreting Partial Derivatives
# ------------------------------

# a) 

# define the function P(C, Y)
def price(C, Y):
    return 16000 + 2400*C - 1800*Y

# calculate price for C = 8 and Y = 5
print(f'Predicted Sales Price: ${price(8,5):,.2f}')

# b)
print('∂P/∂C = 2400')
print('The partial derivative with respect to condition is 2400. For every condition point increase the predicted sales price increases by $2400.')

# c)
print('∂P/∂Y = 1800')
print('The partial derivative with respect to age is -1800. For every one year of age increase the predicted sales price decreases by $1800.')

      
```
