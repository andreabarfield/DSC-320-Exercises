```python
# 1)
# Summarizing Data

# import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# load data
qb = pd.read_csv('qb_stats.csv')

# Set variable for numeric columns
numeric_columns = qb.select_dtypes(include=np.number)

print('Mean of each column: ')
# a) Find the mean of each numerical column of data
for column in numeric_columns.columns:
    data = np.array(qb[column])
    print(f'{column}: {np.mean(data)}')

print('\nStandard Deviation of each column: ')
# b) Find the standard deviation of each numerical column of data
for column in numeric_columns.columns:
    data = np.array(qb[column])
    print(f'{column}: {np.std(data, ddof=1)}')

# c) Create a histogram of the number of yards
# convert yards to array
yards = np.array(qb['yds'])

# create histogram
plt.hist(yards, bins=10, edgecolor='black')

# add labels to histogram
plt.title("Histogram of Passing Yards")
plt.xlabel("Passing Yards")
plt.ylabel("Frequency")

# display histogram
print('')
plt.show()

# d) Create a boxplot for the number of touchdowns
# convert touchdowns to array
touchdowns = np.array(qb['td'])

# create boxplot
plt.boxplot(touchdowns)

# add labels to box plot
plt.title("Boxplots of Touchdowns")
plt.ylabel("Touchdowns")

# display histogram
print('')
plt.show()

```

```python
# 2) Calculating Probabilities from Data

# load the data
survey = pd.read_csv('survey_data.csv')

# a) Probability of college student having brown hair
# Find number of students with brown hair
brown_hair = np.sum(survey['hair_color'] == 'brown')
# Find total number of students
total = len(survey)
# Calculate probability
prob_brown_hair = brown_hair / total
# Display results
print(f'Probability of Brown Hair: {prob_brown_hair}')

# b) Probability of college student having blue eyes
# Find number of students with blue eyes
blue_eyes = np.sum(survey['eye_color'] == 'blue')
# Find total number of students
total = len(survey)
# Calculate probability
prob_blue_eyes = blue_eyes / total
# Display results
print(f'\nProbability of Blue Eyes: {prob_blue_eyes}')

# c) Probability of college student having blue eyes given brown hair
# Filter to only students with brown hair
brown_hair = survey[survey['hair_color'] == 'brown']
# Find probability
blue_given_brown = np.sum(brown_hair['eye_color'] == 'blue') / len(brown_hair)
# Display results
print(f'\nProbability of blue eyes given brown hair: {blue_given_brown}')

# d) Probability of college student have brown hair given blue eyes
# Filter to only students with blue eyes
blue_eyes = survey[survey['eye_color'] == 'blue']
# Calculate probability
brown_given_blue = np.sum(blue_eyes['hair_color'] == 'brown') / len(blue_eyes)
# Display results
print(f'\nProbability of brown hair given blue eyes: {brown_given_blue}')

# e) Based on the data, I can tell that the event of having blue eyes 
# and the event of having brown hair, are independent from one another. 
# I know this because the probabilities of a student having brown hair and
# them having brown hair given blue eyes are very close (.51 and .45). 
# The probabilities of a student having blue eyes and 
# having blue eyes given brown hair is also very close (.25 and .22). 
# Having blue eyes does not change the probability of having brown hair and vice versa.

# f) Bar graph of hair color and eye color 
# Use hair_color column to create an array
hair = np.array(survey['hair_color'])
# Determine options for hair color
hair_colors, counts = np.unique(hair, return_counts=True)
# Create hair bar graph
plt.bar(hair_colors, counts)
plt.title("Hair Color Distribution")
plt.xlabel("Hair Color")
plt.ylabel("Number of Students")
plt.show()

# Use eye_color column to create an array
eye = np.array(survey['eye_color'])
# Determine options for eye color
eye_colors, counts = np.unique(eye, return_counts=True)
# Create eye color bar graph
plt.bar(eye_colors, counts)
plt.title("Eye Color Distribution")
plt.xlabel("Eye Color")
plt.ylabel("Number of Students")
plt.show()
```
