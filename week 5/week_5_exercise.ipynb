```python
# Riemann Sum Function
# Python function that takes in a mathematical function (f),
# two interval endpoints (a & b),
# and the number of triangles (n),
# and outputs the right Riemann sum.

# import numpy
import numpy as np

# define the function that calculates right Riemann sum.
def right_riemann_sum(f, a, b, n):
    delta_x = (b - a) / n
    total = 0 

    # for loop to add together area of all rectangles.
    for k in range(1, n + 1):
        x = a + k * delta_x
        total += f(x) * delta_x

    # return right Riemann sum
    return total

# input function f(x)
def f(x):
    return np.sqrt(4 - x**2)

# calculate right Riemann sum (R4)
R4 = right_riemann_sum(f, 0, 2, 4)
print(f'R4: {R4}')

# continuous interval for loop
for n in [10, 100, 1000]:
    area = right_riemann_sum(f, 0, 2, n)
    print(f'R{n} = {area}')

### As the value of n gets bigger the approximation gets closer to π
```
