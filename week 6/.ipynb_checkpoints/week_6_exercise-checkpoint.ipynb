```python
# import libraries 
import pandas as pd
import numpy as np

# load dataset
housing = pd.read_csv("calif_housing_data.csv")
```

```python
# Part 1 
# Create a Python function that takes in a vector and normalizes it. 

# Define function that normalizes a vector.
def normalize(vector):
    # Find the smallest and largest values. 
    min = np.min(vector)
    max = np.max(vector)
    # Perform normalization on vector.
    normalized_vector = (vector - min) / (max - min)

    return normalized_vector

# Test function by taking a vector in as a numpy array. 
v = np.array([10,20,30,40,50])
print(normalize(v))
```

```python
# Part 2 
# Create a Python function that takes in a vector and standardizes it. 

# Define function that standardizes a vector.
def standardize(vector):
    # Find sample mean.
    samp_mean = np.mean(vector)
    # Find the sample standard deviation
    samp_std = np.std(vector, ddof=1)
    # Perform standardization on vector.
    standardized_vector = (vector - samp_mean) / samp_std

    return standardized_vector

# Test function using vector(v) from part 1.
print(standardize(v))
```

```python
# Part 3
# Working with a dataframe.

# 3a)
# Find number of rows.
print("Number of Rows:", housing.shape[0])

# 3b)
# Create variable "target" to store target vector
target = housing["median_house_value"].to_numpy()
print("Target Vector:", target)

# 3c)
# Create new feature for finding the average number of bedrooms per household.
housing["bedrooms_per_household"] = (
    housing["total_bedrooms"] / housing["households"])
# Print to confirm creation of new column.
housing.head()

# 3d)
housing_features = housing[[
    "housing_median_age",
    "median_income",
    "bedrooms_per_household"
]]
# Print to confirm new dataframe.
housing_features.head()

# 3e)
# Apply standardization to the features 
# using standardize function created in Part 2. 
# Use .apply() function to standardize all columns at once.
housing_features_standardized = housing_features.apply(standardize)

# View results of standardization.
housing_features_standardized.head()

```
